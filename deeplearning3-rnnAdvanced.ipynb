```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('qoute_dataset.csv')
```

```python
df.head()
```

```python
df['quote'][0]
```

```python
quotes = df['quote']
quotes.head()
```

```python
#Preprocess the data
quotes = quotes.str.lower()
```

```python
import string
translator = str.maketrans('','',string.punctuation)
quotes = quotes.apply(lambda x: x.translate(translator))
```

```python
quotes.head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
voc_size = 5000
tok = Tokenizer(num_words=voc_size, oov_token='<OOV>')
tok.fit_on_texts(quotes)
```

```python
word_index  = tok.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
seqs = tok.texts_to_sequences(quotes)
```

```python
print(quotes[1])
print(seqs[1])
```

```python
X = []
y = []

for seq in seqs:
    for i in range(1, len(seq)):
        input_seq = seq[:i]
        output_seq = seq[i]
        X.append(input_seq)
        y.append(output_seq)
```

```python
X
```

```python
y
```

```python
#do padding to make all the input sentences to equal length
maxlen = max(len(x) for x in X)
print(maxlen)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padded  = pad_sequences(X, maxlen = maxlen, padding='pre'
                         )
```

```python
X_padded
```

```python
print(X[0])
print(X_padded[0])
```

```python
y = np.array(y)
```

```python
print(X_padded.shape)
print(y.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, Dense

```

```python
embed_dim = 50
rnn_units = 128
```

```python
rnn_model = Sequential()

rnn_model.add(
    Embedding(
        input_dim=voc_size,
        output_dim=128
    )
)

rnn_model.add(
    SimpleRNN(128)
)

rnn_model.add(
    Dense(voc_size, activation='softmax')
)
```

```python
rnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy']
                 )
```

```python
rnn_model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

lstm_model = Sequential()

lstm_model.add(Embedding(voc_size, output_dim=embed_dim))
lstm_model.add(LSTM(units=rnn_units))
lstm_model.add(Dense(units=voc_size, activation='softmax'))



```

```python
lstm_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)
```

```python
history_rnn = rnn_model.fit(
    X_padded,
    y,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    callbacks=[early_stop]
)
```
